```python
import os

dataset_path = "/content/vegetable_dataset"

print("Dataset folders:", os.listdir(dataset_path))

for split in ["train", "val", "test"]:
    split_path = os.path.join(dataset_path, split)

    print(f"\n{split.upper()}")

    for vegetable in sorted(os.listdir(split_path)):
        vegetable_path = os.path.join(split_path, vegetable)

        if os.path.isdir(vegetable_path):
            print(vegetable, len(os.listdir(vegetable_path)))
```

```python
from ultralytics import YOLO

model = YOLO("yolo11n-cls.pt")

results = model.train(
    data="/content/vegetable_dataset",
    epochs=15,
    imgsz=224,
    batch=16,
    seed=42,
    project="/content/drive/MyDrive/IAT 360 CV PROJECT/training_results",
    name="baseline",
    plots=True
)
```
